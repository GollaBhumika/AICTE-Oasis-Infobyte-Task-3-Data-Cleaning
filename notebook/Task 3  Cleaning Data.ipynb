##  Missing Data Handling

The missing-value treatment is selected separately for each column based on its meaning and data distribution.

- Median imputation may be used for suitable numeric columns because it is less sensitive to extreme values than the mean.
- Mode imputation may be used for categorical columns when the most frequent category is a reasonable replacement.
- Employee identifiers will not be imputed using statistical averages.
- Rows will be removed only when justified by the data quality requirements.

All decisions will be documented after inspecting the actual missing-value patterns.

# Detect outliers using IQR

The IQR method was used to identify potential outliers in Age and Salary. Outliers were retained unless further investigation showed that the values were invalid. This avoids removing legitimate employee records without evidence.

```python
numeric_columns = ["Age", "Salary"]

outlier_results = []

for column in numeric_columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    outlier_results.append({
        "Column": column,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_results)

outlier_report
```
